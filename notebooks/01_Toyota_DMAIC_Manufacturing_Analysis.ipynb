# Toyota Manufacturing Process Improvement Using Lean Six Sigma and Machine Learning

Synthetic educational demonstration; not Toyota factory data.

```python
from src.data_generator import generate_demo_data
from src.preprocessing import clean_data, add_time_features
from src.statistics import baseline_kpis, pareto_by, subgroup_control_data
from src.modeling import train_model
import pandas as pd, matplotlib.pyplot as plt
from pathlib import Path
Path('results').mkdir(exist_ok=True)
```

```python
raw=generate_demo_data(n=5000,seed=42)
display(raw.head())
print(raw.shape, 'synthetic records')
```

```python
clean=clean_data(raw)
featured=add_time_features(clean)
display(baseline_kpis(clean))
display(clean.describe(include='all').T)
```

```python
pareto=pareto_by(clean,'machine_id')
display(pareto)
plt.figure(figsize=(8,4));plt.bar(pareto.machine_id,pareto.defects);plt.title('Synthetic defects by machine');plt.show()
```

```python
model,metrics,cm,features=train_model(featured)
print(metrics,'Confusion matrix:',cm)
```

```python
featured['predicted_defect_probability']=model.predict_proba(featured[features])[:,1]
control=subgroup_control_data(clean)
display(control.head())
```

```python
baseline_kpis(clean).to_csv('results/baseline_kpis.csv',index=False)
pareto.to_csv('results/pareto_defects.csv',index=False)
pd.DataFrame([metrics]).to_csv('results/model_metrics.csv',index=False)
featured.to_csv('results/scored_manufacturing_records.csv',index=False)
control.to_csv('results/control_chart_data.csv',index=False)
print('Saved synthetic demo outputs.')
```
